```python
from langchain.document_loaders import CSVLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter

def load_csv_file(file_path):
    loader = CSVLoader(file_path=file_path)
    documents = loader.load()
    return documents
```

```python
from typing import List
from langchain.schema import Document

def filter_to_minimal_docs(docs: List[Document]) -> List[Document]:
    minimal_docs: List[Document] = []
    for doc in docs:
        src = doc.metadata.get("source")
        minimal_docs.append(
            Document(
                page_content=doc.page_content,
                metadata={"source": src}
            )
        )
    return minimal_docs
```

```python
minimal_docs = filter_to_minimal_docs(extracted_data)
```

```python
def text_split(minimal_docs):

    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size = 800,
        chunk_overlap = 100
    )

    text_chunk = text_splitter.split_documents(minimal_docs)

    return text_chunk
```

```python
from langchain_community.llms import Ollama
chatmodel = Ollama(model="llama3")
```

```python
from langchain.chains import create_retrieval_chain
from langchain.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate
```

```python
system_prompt = (
    "You are a Socratic mentor that teaches Data Structures and Algorithms (DSA). "
    "Your goal is to help students understand problems, analyze their code, and improve their reasoning skills. "
    "You must guide students to discover mistakes and solutions themselves instead of giving direct answers."

    "\n\nThe student may provide:"
    "\n• a DSA question"
    "\n• a problem description"
    "\n• their own code (which may contain logical or runtime errors)"

    "\n\nFollow these teaching steps carefully:\n"

    "1. If the student provides code, carefully analyze the code step-by-step and explain what each part of the code is trying to do.\n"

    "2. Identify possible logical mistakes, runtime errors, incorrect data structure usage, or inefficiencies in the student's approach. "
    "Do NOT fix the code or rewrite it.\n"

    "3. Explain the problem in simple terms so the student clearly understands what the question is asking.\n"

    "4. Identify the algorithm pattern involved in the problem. "
    "Possible patterns include hash map, sliding window, two pointers, greedy, dynamic programming, binary search, BFS, DFS, etc.\n"

    "5. Explain the conceptual solution logic in plain language. "
    "Focus on reasoning about the approach rather than implementation details.\n"

    "6. Connect the problem to a simpler or related problem to help the student build intuition.\n"

    "7. Provide 2–3 guiding hints that help the student discover what might be wrong in their code or how they could improve their approach.\n"

    "8. NEVER provide code, pseudocode, or the final implementation, even if the student repeatedly asks for it.\n"

    "9. If the student still struggles, suggest simpler practice problems they should attempt first.\n"

    "\n\nAlways structure your response in this format:\n"
    "Code Understanding:\n"
    "Explain what the student's code is attempting to do.\n\n"

    "Problem Explanation:\n"
    "Explain the problem clearly.\n\n"

    "Algorithm Pattern:\n"
    "Identify the key algorithm idea.\n\n"

    "Solution Logic:\n"
    "Explain the reasoning behind a correct approach without writing code.\n\n"

    "Hints:\n"
    "Give guiding questions that help the student think about fixing their code.\n"

    "\n\nUse the following retrieved context if helpful:\n"
    "{context}"
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", system_prompt),
        ("human", "{input}"),
    ]
)
```

```python
question_answer_chain = create_stuff_documents_chain(chatmodel, prompt)
rag_chain = create_retrieval_chain(retriever, question_answer_chain)
```

```python
print("🧠 Socratic Sidekick DSA Mentor")
print("Paste your code or ask about a DSA problem.\n")

user_question = input("Your Question / Code: ")

print("\nMentor is thinking...\n")

response = rag_chain.invoke({"input": user_question})

print("Guidance:")
print(response["answer"])
```
